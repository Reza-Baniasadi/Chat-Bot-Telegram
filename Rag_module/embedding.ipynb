```python
#colab version

import pandas as pd
import zipfile
import os

output_folder = "/content/drive/MyDrive/chunks"

os.makedirs(output_folder, exist_ok=True)


zip_path = "/content/drive/MyDrive/kala.zip"
csv_name = "kala.csv"
output_folder = "/content/drive/MyDrive/chunks"

with zipfile.ZipFile(zip_path) as z:
    with z.open(csv_name) as f:
        for i, chunk in enumerate(pd.read_csv(f, chunksize=100000,low_memory=False)):
            chunk['all_information'] = (
                "title_product: " + chunk['title_fa'].astype(str) +
                " rate_people: " + chunk['Rate'].astype(str) +
                " Rate_cnt: " + chunk['Rate_cnt'].astype(str) +
                " Category1: " + chunk['Category1'].astype(str) +
                " category_2: " + chunk['Category2'].astype(str) +
                " Brand: " + chunk['Brand'].astype(str) +
                " price: " + chunk['Price'].astype(str) +
                " advantege: " + chunk['advantages'].astype(str) +
                " disadvantages: " + chunk['disadvantages'].astype(str) +
                " likes: " + chunk['likes'].astype(str) +
                " dislikes: " + chunk['dislikes'].astype(str) +
                " comment_people: " + chunk['comment_text'].astype(str)
            )

            chunk.to_csv(f"{output_folder}/chunk_{i}.csv", index=False)

         
            del chunk


zip_path = "/content/drive/MyDrive/chunks/merge.zip"
csv_filename = "merge.csv"
chunk_size = 10000

chunks = []

with zipfile.ZipFile(zip_path) as z:
    with z.open(csv_filename) as f:
        reader = pd.read_csv(f, chunksize=chunk_size)
        for i, chunk in enumerate(reader):
            chunk = chunk[["all_information"]]
            print(f"Loaded chunk {i}, shape: {chunk.shape}")
            chunks.append(chunk)

df = pd.concat(chunks, ignore_index=True)

print(df.head())

df.to_csv("all_information_only.csv", index=False)
```
