```python
import pandas as pd
import numpy as np
```

```python

data =pd.read_csv('/Users/mac/Documents/Chat-Bot-Telegram/DataSet/Kala/digikala-products.csv')
comment = pd.read_csv("/Users/mac/Documents/Chat-Bot-Telegram/DataSet/Kala/digikala-comments.csv")
data.head(2)
```

```python

data.fillna(' ' , inplace=True)
comment.fillna(' ' , inplace=True)
#data.head(1)
comment.head(20)
```

```python
data = data[data['Price'].notna()]
data['Price']
```

```python
def clean_text(data,column):
    data[column] = data[column].astype(str).str.lower()
    
clean_text(data,'title_fa')      
clean_text(comment,'body')      
clean_text(comment,'title')      
```

```python
data['text'] = " title product : "+ data['title_fa'] + ' Category1 ' + data['Category1'] + ' Category2 ' + data['Category2'] 

merge = pd.merge(data,comment,left_on='id',right_on='product_id',how='left')
merge['comment_text'] = 'id:' + merge['product_id'] + 'title : ' + merge['title'] + ' body : ' + merge['body'] + 'rating :' + merge['rate']

data.fillna(' ' , inplace=True)

data.head(50)
```

```python
data.to_csv('kala.csv')
```
