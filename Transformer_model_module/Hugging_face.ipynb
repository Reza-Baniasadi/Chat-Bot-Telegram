```python
from sentence_transformers import SentenceTransformer
import json
from tqdm import tqdm
import sys
sys.path.append('/Users/mac/Documents/Chat-Bot-Telegram')
from mongoDb_module.mongo import MongoDb
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
db = MongoDb()
```

```python
model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
```

```python
with open('/Users/mac/Documents/Chat-Bot-Telegram/dadrah_dataset1-100000_276342.json','r',encoding="utf-8") as f :
    data= json.load(f)

texts = [item[0]+" "+item[1] for item in data]
#encoding = model.encode(texts)
#print(encoding)

encodefile = np.load('/Users/mac/Documents/Chat-Bot-Telegram/embedding.npy')
```

```python
for text,vectore in zip(texts,encodefile):
    db.insert_question(text,vectore.tolist())
```

```python
all_data = db.get_all_question()

all_vectore = [item['vector'] for item in all_data]
all_questions = [item["question"] for item in all_data]

all_vectore = np.array(all_vectore)


prompt = input('Please enter your question (لطفاً سوال خود را وارد کنید): ')

vector =  model.encode(prompt)
#print([vector])

similatary = cosine_similarity([vector],all_vectore)
#print(similatary)

similatary = np.argmax(similatary)
#print(similatary)

get_question= all_data[similatary]['answer']
print(get_question)




```
